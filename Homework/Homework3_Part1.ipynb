# Homework 3 Part 1: Graph Neural Networks

## Conceptual
1. Why can’t standard neural networks (like MLPs or CNNs) be directly applied to graph‑structured data?
2. What does it mean to “aggregate information from neighbors” in a graph neural network?
3. How does stacking multiple GNN layers change the type of information a node can access?
4. In a GCN, how is information from a node’s neighbors combined?
5. GCNs are often described as “smoothing” node features. What does this mean intuitively, and when might this be harmful? Think about how it handles information from all of its neighbors.
6. What limitation of GCNs motivated the development of Graph Attention Networks?
7. What does “attention” mean in the context of GATs?
8. Conceptually, how does neighborhood aggregation differ between GCNs and GATs?

## Coding

In this assignment, you will apply Graph Neural Networks (GNNs) to a real social network by performing node classification on the Facebook Page–Page dataset, where nodes represent verified Facebook pages, edges represent mutual likes between pages, and labels correspond to page categories (e.g., politician, company, media). Using PyTorch Geometric, load the dataset via torch_geometric.datasets.FacebookPagePage, which automatically downloads and processes the data (official documentation: https://pytorch-geometric.readthedocs.io/en/latest/generated/torch_geometric.datasets.FacebookPagePage.html), or alternatively download the raw files from the UCI Machine Learning Repository (https://archive.ics.uci.edu/dataset/527/facebook+large+page+page+network). You should train and evaluate both a Graph Convolutional Network (GCN) and a Graph Attention Network (GAT) to predict node labels, compare their test accuracies, and briefly discuss how fixed neighbor aggregation (GCN) differs conceptually from attention‑based aggregation (GAT) in the context of social networks.
